```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("Dataset preprocessing.csv")

for col in df.select_dtypes(include='number').columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outlier_count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_percent = (outlier_count / len(df)) * 100

    print(col, outlier_count, round(outlier_percent, 2), "%")
    
```

```python
import pandas as pd
df = pd.read_csv("Dataset preprocessing.csv")

# 1. Ratios
df["Income_per_Age"] = df["Annual_Income"] / df["Age"]
df["Spend_per_Income"] = df["Spending_Score"] / df["Annual_Income"]
print(df.head())
```

```python
# 2. Aggregated columns
df["Total_Purchase"] = df[["Spending_Score", "Purchase_Frequency", "Last_Purchase_Amount"]].sum(axis=1)
df["Average_Purchase"] = df[["Spending_Score", "Purchase_Frequency", "Last_Purchase_Amount"]].mean(axis=1)
print(df.head())
```

```python
# 3. Derived features
df["Age_Group"] = df["Age"] // 10   
df["Score_per_Age"] = df["Spending_Score"] / df["Age"]
print(df.head())
```

```python
#12.
import pandas as pd
from sklearn.feature_selection import VarianceThreshold
df = pd.read_csv("Dataset preprocessing.csv")

# Remove constant features
for col in df.columns:
    if df[col].nunique() == 1:
        df.drop(col, axis=1, inplace=True)
print(df.columns)
```

```python
num_df = df.select_dtypes(include=['int64','float64'])
# Remove low variance features
selector = VarianceThreshold(threshold=0.01)
selector.fit(num_df)
```

```python
# Final selected features
selected_features = num_df.columns[selector.get_support()]

print("Final Selected Features:")
print(selected_features)
```

```python
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler

df = pd.read_csv("Dataset preprocessing.csv")

# Handle missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

for col in df.select_dtypes(include='object'):
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
# Encode categorical columns
le = LabelEncoder()

for col in df.select_dtypes(include='object'):
    df[col] = le.fit_transform(df[col])
```

```python
# Scale numerical features
scaler = StandardScaler()

df[df.columns] = scaler.fit_transform(df)

print(df.head())
```
